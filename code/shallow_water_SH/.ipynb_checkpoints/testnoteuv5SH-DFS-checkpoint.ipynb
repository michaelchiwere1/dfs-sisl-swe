```python
#Plotting the evolution of the errors

#TIMESTEPS
NN = np.array([3600/i for i in [1, 2, 4, 8, 16, 32, 64, 128]])


fig, (ax1, ax2,  ax3) = plt.subplots(1, 3, figsize=(20, 6))

ax1.loglog(NN , ss[:,0],'b.-',label="Proposed Method")
ax1.loglog(NN, ss1[:,0],'rd-', label='Current Method')

ax1.legend()
ax1.set_title("L1-norm of  the Error")
ax1.set_ylabel("Error")
ax1.set_xlabel("N")

ax2.loglog(NN, ss[:,1],'b.-',label="Proposed Method")
ax2.loglog(NN, ss1[:,1],'rd-', label='Current Method')
ax2.legend()
ax2.set_title("L2-norm of  the Error")
ax2.set_ylabel("Error")
ax2.set_xlabel("N")

ax3.loglog(NN, ss[:,2],'b.-',label="Proposed Method")
ax3.loglog(NN, ss1[:,2],'rd-', label='Current Method')
ax3.legend()
ax3.set_title("Max-norm of  the Error")
ax3.set_ylabel("Error")
ax3.set_xlabel("N")
plt.rcParams.update({'font.size': 18})
fig.tight_layout()
plt.show()
```
